```python
import pandas as pd

df = pd.read_csv("/content/sample_data/Book1.csv")
df.head()
```

```python
import matplotlib.pyplot as plt


plt.scatter(df['Distance (km)'], df['Delivery Time (minutes)'], color="#fff", edgecolor="k")
plt.xlabel("Distance in Km")
plt.ylabel("Delivery Time in minutes")
plt.title("Farther Delivery take more time.")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(df[['Distance (km)']], df['Delivery Time (minutes)']) # here we are training the model
```

```python
model.predict([[1]])
```

```python
model.coef_ # m = slope(Gradient)
```

```python
model.intercept_ # b = intercept
```

```python
# So we know that y = m * x + b
# time = m * distance + b
time = 5.74 * 2 + 6.13
time
```

```python
model.predict([[2]])
# So the models predictions value is calculated through the formula y = m * x + b
```

```python
plt.scatter(df['Distance (km)'], df['Delivery Time (minutes)'], color="#fff", edgecolor="k")
plt.plot(df['Distance (km)'], model.predict(df [['Distance (km)']]), color='#39D353', linewidth=2)
plt.xlabel("Distance (km)")
plt.ylabel("Delivery Time (min)")
plt.title("Best fit line: time m distance + b")
plt.show()
```
