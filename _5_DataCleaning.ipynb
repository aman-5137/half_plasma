```python
# To find duplicate rows
duplicate_rows = df[df.duplicated(keep = False)]
print(duplicate_rows)
```

```python
before_duplicates_removed = df.shape[0]
df = df.drop_duplicates()
after_duplicates_removed = df.shape[0]

print("Before removing duplicates: ", before_duplicates_removed)
print("After removing duplicates: ", after_duplicates_removed)
```

```python
# We do not need unnecessary columns during a model training so we will remove them too
df = df.drop(columns=['applicant_id', 'name'])
```

```python
# Clean the 'income' column by removing '$' and ',' characters and converting to numeric
df['income'] = df['income'].astype(str).str.replace(r'[$ ,]', '', regex=True)
df['income'] = pd.to_numeric(df['income'], errors='coerce')
```

```python
df['gender'] = df['gender'].map({"male":"Male", 'm': "Male", "female": "Female", "f" : "Female"})
```
